```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
import os

os.listdir()
```

```python
daily = pd.read_csv("pharma_sales_daily.csv")
monthly = pd.read_csv("pharma_sales_monthly.csv")
```

```python
display(daily.head())
display(monthly.head())
```

```python
print("Daily columns:", daily.columns.tolist())
print("Monthly columns:", monthly.columns.tolist())
```

```python
#rename the columns 
#1.Create a column-renaming dictionary
rename_map = {
    "datum": "sales_date",
    "M01AB": "m01ab_sales",
    "M01AE": "m01ae_sales",
    "N02BA": "n02ba_sales",
    "N02BE": "n02be_sales",
    "N05B": "n05b_sales",
    "N05C": "n05c_sales",
    "R03": "r03_sales",
    "R06": "r06_sales",
    "Year": "year",
    "Hour": "hour",
    "Weekday Name": "sales_weekday",
    "Month": "sales_month"
}
#applying the names to column
daily = daily.rename(columns=rename_map)
monthly = monthly.rename(columns=rename_map)

print("Daily columns:", daily.columns.tolist())
print("Monthly columns:", monthly.columns.tolist())
```

```python
daily.info()
monthly.info()
```

```python
#Convert the date column to datetime
daily["sales_date"] = pd.to_datetime(
    daily["sales_date"], errors="coerce"
)

monthly["sales_date"] = pd.to_datetime(
    monthly["sales_date"], errors="coerce"
)
print(daily["sales_date"].dtype)
print(monthly["sales_date"].dtype)

#checking for invalid dates
print("Invalid daily dates:", daily["sales_date"].isna().sum())
print("Invalid monthly dates:", monthly["sales_date"].isna().sum())
```

```python
#checking for null values
print("Daily missing values:")
print(daily.isnull().sum())

print("\nMonthly missing values:")
print(monthly.isnull().sum())
```

```python
#Checking duplicate records
print("Daily duplicate rows:", daily.duplicated().sum())
print("Monthly duplicate rows:", monthly.duplicated().sum())

print("Duplicate daily dates:", daily["sales_date"].duplicated().sum())
print("Duplicate monthly dates:", monthly["sales_date"].duplicated().sum())
```

```python
#sort records by date 
daily = daily.sort_values(
    "sales_date"
).reset_index(drop=True)

monthly = monthly.sort_values(
    "sales_date"
).reset_index(drop=True)
```

```python
sales_cols = [
    "m01ab_sales",
    "m01ae_sales",
    "n02ba_sales",
    "n02be_sales",
    "n05b_sales",
    "n05c_sales",
    "r03_sales",
    "r06_sales"
]
daily["total_sales_volume"] = daily[sales_cols].sum(
    axis=1, min_count=len(sales_cols)
)

monthly["total_sales_volume"] = monthly[sales_cols].sum(
    axis=1, min_count=len(sales_cols)
)
```

```python
daily.to_csv(
    r"D:\New folder\PHARMA SALES ANALYSIS\Dataset\Cleaned\pharma_sales_daily_cleaned.csv",
    index=False
)

monthly.to_csv(
    r"D:\New folder\PHARMA SALES ANALYSIS\Dataset\Cleaned\pharma_sales_monthly_cleaned.csv",
    index=False
)
```

```python
#Analyze total sales by drug category

category_sales = (
    monthly[sales_cols]
    .sum()
    .sort_values(ascending=False)
)

display(category_sales)
```

```python

##Which drug categories contribute the most to total sales volume?
category_sales.plot(
    kind="bar",
    figsize=(10, 5),
    title="Total Sales Volume by Drug Category"
)

plt.ylabel("Total Sales Volume")
plt.xlabel("Drug Category")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
#Is total monthly sales volume increasing, decreasing, or fluctuating over time
plt.figure(figsize=(12, 5))

plt.plot(
    monthly["sales_date"],
    monthly["total_sales_volume"]
)

plt.title("Monthly Pharmaceutical Sales Volume")
plt.xlabel("Month")
plt.ylabel("Total Sales Volume")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
#Are there unusual spikes or drops in daily sales?
plt.figure(figsize=(12, 5))

plt.plot(
    daily["sales_date"],
    daily["total_sales_volume"],
    linewidth=0.8
)

plt.title("Daily Pharmaceutical Sales Volume")
plt.xlabel("Date")
plt.ylabel("Total Sales Volume")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
#highest and lowest sales days
highest_day = daily.loc[
    daily["total_sales_volume"].idxmax()
]

lowest_day = daily.loc[
    daily["total_sales_volume"].idxmin()
]

print("Highest sales day:")
print(highest_day[["sales_date", "total_sales_volume"]])

print("\nLowest sales day:")
print(lowest_day[["sales_date", "total_sales_volume"]])
```
